```python
import pandas as pd 
import seaborn as sns
import numpy as np
df = pd.read_excel("Gold price.xlsx")
df.head()
```

```python
import matplotlib.pyplot as plt
plt.plot(df["Date"], df["Price"])
```

```python
df['date']=pd.to_datetime(df['Date'])
df=df.sort_values('date')
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
```

```python
yearly_avg = df.groupby('Year')['Price'].mean()
print(yearly_avg)
```

```python
yearly_growth = yearly_avg.pct_change()*100
print(yearly_growth)
```

```python
plt.figure()
plt.plot(df['Date'], df['Price'])
plt.title("Gold Price Trend (2021-2025)")
plt.xlabel("Date")
plt.ylabel("Price")
plt.show()
```

```python
monthly_avg = df.groupby('Month')['Price'].mean()
print(monthly_avg)
```

```python
df.head()
df.tail()
df.info()
df.describe()
df.shape
df.columns
```

```python
df.isnull().sum()
df.dropna()
df.fillna(0)
```

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

```python
df.dtypes
df['Price'] = df['Price'].astype(float)
```

```python

import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
plt.plot(df['Date'], df['Price'], marker='o')   
plt.title("Gold Price Trend (2021-2025)")
plt.xlabel("Date")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
plt.figure(figsize=(10, 6))
sns.lineplot(x='Date', y='Price', data=df, marker='o')
plt.title("Gold Price Trend (2021-2025)")   
plt.xlabel("Date")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
```

```python
yearly_avg = df.groupby('Year')['Price'].mean()
plt.figure(figsize=(8, 5))
yearly_avg.plot(kind='bar')
plt.title("Average Gold Price by Year")
plt.xlabel("Year")
plt.ylabel("Average Price")
plt.show()
```

```python
yearly_avg = df.groupby('Year')['Price'].mean()
plt.figure(figsize=(8, 5))
yearly_avg.plot(kind='bar')
plt.title("Average Gold Price by Year")
plt.xlabel("month")
plt.ylabel("Average Price")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Price'], bins=20, kde=True)
plt.title("Distribution of Daily Price Range")
plt.xlabel("Daily Range")
plt.ylabel("Frequency")
plt.show()
```
